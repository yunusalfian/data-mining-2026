## Hypothesis Testing dengan Dataset Titanic

Notebook ini merupakan latihan *Hypothesis Testing* menggunakan dataset Titanic. Kita akan menguji apakah **tingkat keselamatan penumpang berbeda berdasarkan jenis kelamin**.

**Dataset:** `data/titanic.csv` (gunakan file `train.csv` dari Kaggle Titanic, lalu ubah namanya menjadi `titanic.csv`).

**Tujuan pembelajaran:**
- Memahami Null Hypothesis (H₀) dan Alternative Hypothesis (H₁).
- Menentukan tingkat signifikansi (α).
- Menghitung statistik uji dan p-value.
- Menarik kesimpulan berdasarkan hasil pengujian.
- Memahami Type I Error dan keterbatasan hasil pengujian.

## 1. Import Library

Kita menggunakan Pandas untuk mengolah data, Matplotlib untuk visualisasi, dan SciPy untuk melakukan pengujian statistik.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

%matplotlib inline
````

## 2. Membaca Dataset

Pastikan struktur folder seperti berikut:

```text
data-mining-weekly/
├── data/
│   └── titanic.csv
└── titanic_hypothesis_testing.ipynb
```

````python
df = pd.read_csv("data/titanic.csv")
df.head()
````

## 3. Memahami Data

````python
print("Ukuran dataset:", df.shape)
print("\\nNama kolom:")
print(df.columns.tolist())
print("\\nInformasi dataset:")
df.info()
````

````python
# Periksa jumlah data kosong pada kolom yang akan digunakan
df[["Sex", "Survived"]].isnull().sum()
````

## 4. Menyiapkan Data

Kolom `Survived` berisi:
- `0` = tidak selamat
- `1` = selamat

Kolom `Sex` berisi jenis kelamin penumpang. Kita hanya menggunakan dua kolom ini untuk pengujian.

````python
data = df[["Sex", "Survived"]].dropna().copy()
data["Sex"].value_counts()
````

## 5. Eksplorasi Awal

Hitung jumlah penumpang dan rata-rata keselamatan berdasarkan jenis kelamin. Rata-rata kolom `Survived` dapat dibaca sebagai proporsi penumpang yang selamat.

````python
summary = data.groupby("Sex")["Survived"].agg(["count", "sum", "mean"])
summary = summary.rename(columns={
    "count": "Jumlah Penumpang",
    "sum": "Jumlah Selamat",
    "mean": "Tingkat Keselamatan"
})
summary
````

````python
summary["Tingkat Keselamatan"].plot(
    kind="bar",
    ylim=(0, 1),
    ylabel="Proporsi Selamat",
    xlabel="Jenis Kelamin",
    title="Tingkat Keselamatan Berdasarkan Jenis Kelamin",
    rot=0
)
plt.show()
````

## 6. Menentukan Hipotesis

Pertanyaan penelitian: **Apakah proporsi penumpang yang selamat berbeda antara perempuan dan laki-laki?**

- **H₀ (Null Hypothesis):** Proporsi keselamatan perempuan dan laki-laki sama.
- **H₁ (Alternative Hypothesis):** Proporsi keselamatan perempuan dan laki-laki berbeda.

Kita menggunakan tingkat signifikansi **α = 0,05 (5%)**. Pengujian ini dua arah (*two-sided*) karena kita menguji apakah proporsinya berbeda, bukan menentukan arah perbedaannya.

````python
alpha = 0.05

female = data[data["Sex"] == "female"]["Survived"]
male = data[data["Sex"] == "male"]["Survived"]

female_survived = female.sum()
female_total = female.count()
male_survived = male.sum()
male_total = male.count()

print("Perempuan:", female_survived, "selamat dari", female_total)
print("Laki-laki:", male_survived, "selamat dari", male_total)
print("Proporsi selamat perempuan:", round(female.mean(), 3))
print("Proporsi selamat laki-laki:", round(male.mean(), 3))
````

## 7. Melakukan Chi-Square Test

Kedua variabel yang diuji bersifat kategorikal:
- `Sex`: female atau male
- `Survived`: 0 atau 1

Karena itu, kita menggunakan **Chi-Square Test of Independence** untuk menguji apakah jenis kelamin dan status keselamatan saling berhubungan.

````python
contingency_table = pd.crosstab(data["Sex"], data["Survived"])
contingency_table
````

````python
chi2, p_value, degrees_of_freedom, expected = stats.chi2_contingency(contingency_table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", degrees_of_freedom)
print("\\nExpected frequencies:")
print(pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
).round(2))
````

## 8. Mengambil Keputusan

Aturan keputusan:
- Jika **p-value < 0,05**, tolak H₀.
- Jika **p-value ≥ 0,05**, gagal menolak H₀.

Ingat: gagal menolak H₀ bukan berarti H₀ terbukti benar.

````python
if p_value < alpha:
    print("Keputusan: Tolak H0")
    print("Kesimpulan: Terdapat hubungan yang signifikan secara statistik antara jenis kelamin dan keselamatan.")
else:
    print("Keputusan: Gagal menolak H0")
    print("Kesimpulan: Belum terdapat bukti yang cukup bahwa jenis kelamin dan keselamatan berhubungan.")
````

## 9. Memahami Type I Error dan Type II Error

- **Type I Error:** Menolak H₀ padahal H₀ sebenarnya benar. Dalam contoh ini, kita menyimpulkan ada hubungan antara jenis kelamin dan keselamatan padahal sebenarnya tidak ada.
- **Type II Error:** Gagal menolak H₀ padahal H₀ sebenarnya salah. Dalam contoh ini, kita tidak menemukan hubungan meskipun hubungan tersebut sebenarnya ada.

Dengan α = 0,05, tingkat risiko Type I Error yang ditetapkan sebelum pengujian adalah 5%.

## 10. Latihan Tambahan: Apakah Usia Penumpang yang Selamat Berbeda?

Sekarang lakukan pengujian tambahan untuk membandingkan usia penumpang yang selamat dan tidak selamat.

- **H₀:** Rata-rata usia kedua kelompok sama.
- **H₁:** Rata-rata usia kedua kelompok berbeda.

Kita menggunakan Welch's t-test (`equal_var=False`) karena varians kedua kelompok tidak harus dianggap sama. Baris dengan usia kosong dihilangkan hanya untuk pengujian ini.

````python
age_data = df[["Age", "Survived"]].dropna()

age_survived = age_data[age_data["Survived"] == 1]["Age"]
age_not_survived = age_data[age_data["Survived"] == 0]["Age"]

print("Rata-rata usia penumpang selamat:", round(age_survived.mean(), 2))
print("Rata-rata usia penumpang tidak selamat:", round(age_not_survived.mean(), 2))
````

````python
t_stat, age_p_value = stats.ttest_ind(
    age_survived,
    age_not_survived,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", age_p_value)

if age_p_value < alpha:
    print("Tolak H0: terdapat perbedaan rata-rata usia yang signifikan.")
else:
    print("Gagal menolak H0: belum ada bukti perbedaan rata-rata usia yang signifikan.")
````
