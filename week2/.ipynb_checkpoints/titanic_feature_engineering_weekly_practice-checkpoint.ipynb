# Data Mining Weekly Practice
## Feature Engineering — Titanic Dataset

**Tujuan:** memahami cara membuat dan mengubah fitur dari dataset Titanic setelah melakukan EDA.

Dataset: `titanic.csv` dari Kaggle Titanic (`train.csv` yang diunduh lalu diganti nama).

**Alur:** memeriksa data → membuat fitur baru → mengubah fitur kategorikal → melihat hasilnya.

> Praktikum ini berfokus pada Feature Engineering, bukan melatih model Machine Learning.

## 1. Import Library

Kita menggunakan Pandas untuk mengolah data dan Matplotlib untuk melihat distribusi fitur.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Memuat Dataset

Simpan file `titanic.csv` di folder `data`, sehingga lokasinya menjadi `data/titanic.csv`.

```python
titanic = pd.read_csv('data/titanic.csv')
titanic.head()
```

## 3. Periksa Kolom yang Dibutuhkan

Pilih beberapa kolom yang relevan untuk contoh feature engineering. `SibSp` adalah jumlah saudara/pasangan di kapal, sedangkan `Parch` adalah jumlah orang tua/anak di kapal.

```python
titanic[['Survived', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']].head()
```

## 4. Periksa Nilai Kosong

Nilai kosong perlu dikenali sebelum fitur dipakai. Untuk contoh ini, kita isi `Age` yang kosong dengan median.

```python
titanic[['Age', 'Embarked', 'Fare']].isnull().sum()
```

```python
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])
```

## 5. Membuat Fitur Baru: `FamilySize`

Kita gabungkan `SibSp` dan `Parch`, lalu tambahkan 1 untuk menghitung penumpang itu sendiri. Fitur baru ini merangkum ukuran keluarga yang bepergian bersama.

```python
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch'] + 1
titanic[['SibSp', 'Parch', 'FamilySize']].head(10)
```

## 6. Membuat Fitur Biner: `IsAlone`

Fitur ini bernilai 1 jika penumpang bepergian sendirian, dan 0 jika tidak.

```python
titanic['IsAlone'] = (titanic['FamilySize'] == 1).astype(int)
titanic[['FamilySize', 'IsAlone']].head(10)
```

## 7. Mengubah Fitur Kategorikal dengan One-Hot Encoding

Kolom `Sex` berisi kategori teks. One-hot encoding mengubah kategori menjadi kolom numerik, misalnya `Sex_male` dan `Sex_female`.

```python
titanic_encoded = pd.get_dummies(
    titanic,
    columns=['Sex', 'Embarked'],
    dtype=int
)
titanic_encoded[['Sex_female', 'Sex_male', 'Embarked_C', 'Embarked_Q', 'Embarked_S']].head()
```

## 8. Membuat Kelompok Usia

Kadang usia lebih mudah dianalisis sebagai kelompok, bukan angka mentah. Di sini kita membuat kategori usia sederhana.

```python
titanic['AgeGroup'] = pd.cut(
    titanic['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'],
    include_lowest=True
)
titanic[['Age', 'AgeGroup']].head(10)
```

## 9. Melihat Hubungan Fitur Baru dengan Keselamatan

`Survived` adalah label yang menunjukkan apakah penumpang selamat (1) atau tidak (0). Kita bandingkan rata-rata keselamatan berdasarkan ukuran keluarga. Ini hanya eksplorasi, bukan bukti sebab-akibat.

```python
titanic.groupby('FamilySize', observed=True)['Survived'].mean().round(3)
```

```python
titanic.groupby('IsAlone', observed=True)['Survived'].mean().plot(kind='bar')
plt.title('Rata-rata Survival: Sendirian vs Bersama Keluarga')
plt.xlabel('IsAlone (1 = sendirian)')
plt.ylabel('Rata-rata Survived')
plt.xticks(rotation=0)
plt.show()
```

## 10. Simpan Hasil

Kita simpan dataset yang sudah memiliki fitur baru. File hasil akan dibuat di folder kerja notebook.

```python
titanic.to_csv('titanic_feature_engineered.csv', index=False)
print('File tersimpan: titanic_feature_engineered.csv')
```

## Kesimpulan

Feature Engineering adalah proses membuat atau mengubah kolom agar lebih berguna untuk analisis atau Machine Learning.

Dalam praktikum ini:
- `FamilySize` menggabungkan jumlah saudara/pasangan dan orang tua/anak.
- `IsAlone` menandai apakah penumpang bepergian sendirian.
- `AgeGroup` mengelompokkan usia.
- One-hot encoding mengubah kategori teks menjadi kolom numerik.
