Logistic Regression dan Evaluasi Model - Dataset WineQT

Dataset `WineQT.csv` berisi pengukuran karakteristik kimia wine. Kolom `quality` menjadi target yang akan diprediksi. Kolom `Id` hanya berfungsi sebagai identitas baris sehingga tidak digunakan sebagai fitur model.

Mengimpor library

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score
)

%matplotlib inline
sns.set_theme(style="whitegrid")
```

Memuat dan memahami data


```python
data = pd.read_csv("data/WineQT.csv")

print("Ukuran dataset:", data.shape)
display(data.head())
display(data.info())
```

```python
display(data.dtypes.value_counts().to_frame("Jumlah Kolom"))
display(data.isnull().sum().to_frame("Missing Values"))
display(data.describe().T)
```

Menentukan fitur dan target

Kolom `quality` digunakan sebagai target (y). Kolom `Id` dikeluarkan karena bukan karakteristik kimia wine. Kolom lainnya menjadi fitur (X).

```python
data = data.drop(columns=["Id"], errors="ignore")

X = data.drop(columns=["quality"])
y = data["quality"]

print("Jumlah fitur:", X.shape[1])
print("Nama fitur:", X.columns.tolist())
print("Kategori target:", sorted(y.unique()))
```

Memeriksa distribusi kelas

Distribusi kelas perlu diperiksa sebelum pelatihan. Pada dataset ini jumlah sampel untuk setiap kualitas tidak sama. Ketidakseimbangan kelas dapat membuat model lebih mudah mengenali kelas mayoritas dibandingkan kelas yang jumlahnya sedikit.

```python
class_counts = y.value_counts().sort_index()
display(class_counts.to_frame("Jumlah Sampel"))

plt.figure(figsize=(8, 4))
sns.countplot(x=y, order=class_counts.index)
plt.title("Distribusi Kualitas Wine")
plt.xlabel("Quality")
plt.ylabel("Jumlah Sampel")
plt.show()
```

Encoding target

Nilai target pada dataset sudah berupa angka, tetapi kita dapat menggunakan `LabelEncoder` agar proses pemetaan kelas dilakukan secara eksplisit. Encoder menyimpan hubungan antara label asli dan label numerik.

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

class_mapping = pd.DataFrame({
    "Label Asli": label_encoder.classes_,
    "Label Encoded": range(len(label_encoder.classes_))
})
display(class_mapping)
```

Melihat korelasi antarfitur

Korelasi menunjukkan hubungan linear antarfitur. Nilai mendekati +1 menunjukkan hubungan positif yang kuat, sedangkan nilai mendekati -1 menunjukkan hubungan negatif yang kuat. Korelasi tinggi antarfitur dapat mengindikasikan multikolinearitas.

```python
corr_matrix = X.corr(numeric_only=True)

plt.figure(figsize=(12, 9))
sns.heatmap(corr_matrix, cmap="coolwarm", center=0)
plt.title("Matriks Korelasi Fitur Wine")
plt.tight_layout()
plt.show()
```

```python
#Mengambil pasangan korelasi tanpa duplikasi diagonal
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)
corr_pairs = (
    upper_triangle.stack()
    .rename("correlation")
    .reset_index()
    .rename(columns={"level_0": "feature_1", "level_1": "feature_2"})
)
corr_pairs["absolute_correlation"] = corr_pairs["correlation"].abs()

display(
    corr_pairs.sort_values("absolute_correlation", ascending=False).head(10)
)
```

## 8. Membagi data training dan testing

`StratifiedShuffleSplit` membagi data dengan mempertahankan proporsi kelas target pada data training dan testing. Data testing disimpan untuk evaluasi akhir dan tidak digunakan saat pelatihan model.

```python
splitter = StratifiedShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, test_idx = next(splitter.split(X, y_encoded))

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()
y_train = y_encoded[train_idx]
y_test = y_encoded[test_idx]

print("Data training:", X_train.shape)
print("Data testing:", X_test.shape)
```

```python
# Membandingkan proporsi kelas pada data awal, training, dan testing
distribution = pd.DataFrame({
    "Keseluruhan": pd.Series(y_encoded).value_counts(normalize=True).sort_index(),
    "Training": pd.Series(y_train).value_counts(normalize=True).sort_index(),
    "Testing": pd.Series(y_test).value_counts(normalize=True).sort_index()
})
distribution.index = label_encoder.classes_
display(distribution)
```

Standardisasi fitur

Fitur kimia memiliki skala dan rentang nilai yang berbeda. Logistic Regression dapat lebih stabil ketika fitur distandardisasi. `StandardScaler` mengubah fitur agar memiliki rata-rata mendekati 0 dan standar deviasi mendekati 1.

Standardisasi ditempatkan di dalam `Pipeline`. Dengan cara ini scaler hanya mempelajari parameter dari data training pada setiap proses validasi silang, sehingga menghindari data leakage.

Membangun model Logistic Regression

Tiga model akan dibandingkan:

1. Tanpa regularisasi - model dasar.
2. L1 regularization - dapat mengecilkan sebagian koefisien hingga nol.
3. L2 regularization - mengecilkan koefisien agar model tidak terlalu kompleks.

Nilai parameter regularisasi untuk model L1 dan L2 dipilih melalui cross-validation. Solver `saga` mendukung regularisasi L1 dan L2 serta klasifikasi multiclass.

```python
# Model dasar tanpa regularisasi
model_lr = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        penalty=None,
        solver="lbfgs",
        max_iter=5000
    ))
])

model_lr.fit(X_train, y_train)
```

```python
# Model dengan regularisasi L1 dan pencarian parameter melalui cross-validation
model_l1 = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegressionCV(
        Cs=10,
        cv=5,
        penalty="l1",
        solver="saga",
        scoring="accuracy",
        max_iter=10000,
        random_state=42
    ))
])

model_l1.fit(X_train, y_train)
```

```python
# Model dengan regularisasi L2 dan pencarian parameter melalui cross-validation
model_l2 = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegressionCV(
        Cs=10,
        cv=5,
        penalty="l2",
        solver="lbfgs",
        scoring="accuracy",
        max_iter=5000
    ))
])

model_l2.fit(X_train, y_train)
```

Membandingkan koefisien model

Koefisien menunjukkan arah dan besarnya hubungan fitur dengan skor keputusan kelas. Besar koefisien perlu dibaca dengan mempertimbangkan bahwa fitur telah distandardisasi.

Pada klasifikasi multiclass, model menghasilkan koefisien untuk setiap kelas. Regularisasi L1 cenderung menghasilkan lebih banyak koefisien bernilai nol dibandingkan L2.

```python
model_dict = {
    "Logistic Regression": model_lr,
    "L1": model_l1,
    "L2": model_l2
}

coefficient_frames = {}
for name, pipeline in model_dict.items():
    classifier = pipeline.named_steps["classifier"]
    coefficient_frames[name] = pd.DataFrame(
        classifier.coef_.T,
        index=X.columns,
        columns=label_encoder.classes_
    )

display(coefficient_frames["Logistic Regression"].head())
```

```python
#Visualisasi jumlah koefisien nol pada tiap model
zero_counts = {}
for name, pipeline in model_dict.items():
    coef = pipeline.named_steps["classifier"].coef_
    zero_counts[name] = int(np.isclose(coef, 0).sum())

display(pd.Series(zero_counts, name="Jumlah Koefisien Mendekati Nol"))
```

```python
#Menampilkan koefisien setiap model untuk seluruh kelas
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

for ax, (name, coef_df) in zip(axes, coefficient_frames.items()):
    sns.heatmap(coef_df, cmap="coolwarm", center=0, ax=ax)
    ax.set_title(name)
    ax.set_xlabel("Kelas Quality")
    ax.set_ylabel("Fitur")

plt.tight_layout()
plt.show()
```

Melakukan prediksi

Setiap model digunakan untuk memprediksi kelas pada data testing. Selain label prediksi, `predict_proba()` menghasilkan probabilitas untuk setiap kelas. Probabilitas tertinggi menjadi kelas prediksi model.

```python
predictions = {}
probabilities = {}

for name, model in model_dict.items():
    predictions[name] = model.predict(X_test)
    probabilities[name] = model.predict_proba(X_test)

prediction_comparison = pd.DataFrame({
    "Aktual": label_encoder.inverse_transform(y_test),
    **{
        name: label_encoder.inverse_transform(pred)
        for name, pred in predictions.items()
    }
})

display(prediction_comparison.head(10))
```

```python
# Probabilitas tertinggi untuk setiap prediksi
probability_comparison = pd.DataFrame({
    name: prob.max(axis=1)
    for name, prob in probabilities.items()
})
display(probability_comparison.head())
```

Evaluasi model dengan metrik klasifikasi

Evaluasi dilakukan menggunakan Accuracy, Precision, Recall, F1-score, dan ROC-AUC. Untuk Precision, Recall, dan F1-score digunakan rata-rata **weighted**, sehingga kontribusi setiap kelas mempertimbangkan jumlah sampelnya.

ROC-AUC dihitung dengan pendekatan One-vs-Rest (OvR) menggunakan probabilitas prediksi. Untuk dataset multiclass, penggunaan probabilitas lebih tepat daripada menghitung AUC dari label kelas hasil prediksi.

```python
results = []

for name, model in model_dict.items():
    y_pred = predictions[name]
    y_prob = probabilities[name]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision (weighted)": precision_score(
            y_test, y_pred, average="weighted", zero_division=0
        ),
        "Recall (weighted)": recall_score(
            y_test, y_pred, average="weighted", zero_division=0
        ),
        "F1-score (weighted)": f1_score(
            y_test, y_pred, average="weighted", zero_division=0
        ),
        "ROC-AUC (OvR)": roc_auc_score(
            y_test, y_prob, multi_class="ovr", average="weighted"
        )
    })

metrics_df = pd.DataFrame(results).set_index("Model")
display(metrics_df.round(4))
```

Classification report

Classification report menampilkan Precision, Recall, F1-score, dan support untuk masing-masing kelas. Bagian ini penting karena nilai metrik keseluruhan saja dapat menyembunyikan performa yang rendah pada kelas minoritas.

```python
for name, y_pred in predictions.items():
    print(f"\n=== {name} ===")
    print(classification_report(
        y_test,
        y_pred,
        labels=np.arange(len(label_encoder.classes_)),
        target_names=[str(c) for c in label_encoder.classes_],
        zero_division=0
    ))
```

Confusion Matrix

Confusion matrix memperlihatkan jumlah prediksi benar dan salah untuk setiap kelas. Baris menunjukkan kelas aktual, sedangkan kolom menunjukkan kelas hasil prediksi.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, (name, y_pred) in zip(axes, predictions.items()):
    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=np.arange(len(label_encoder.classes_))
    )
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_,
        ax=ax
    )
    ax.set_title(name)
    ax.set_xlabel("Prediksi")
    ax.set_ylabel("Aktual")

plt.tight_layout()
plt.show()
```

Membandingkan performa model

Tabel dan grafik berikut memudahkan perbandingan performa ketiga model. Perhatikan bahwa Accuracy yang tinggi belum tentu berarti seluruh kelas diprediksi dengan baik. Periksa juga F1-score weighted, classification report, dan confusion matrix.

```python
metrics_df[["Accuracy", "Precision (weighted)", "Recall (weighted)", "F1-score (weighted)", "ROC-AUC (OvR)"]].plot(
    kind="bar",
    figsize=(12, 5),
    ylim=(0, 1),
    rot=0
)
plt.title("Perbandingan Performa Logistic Regression")
plt.ylabel("Skor")
plt.xlabel("Model")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```
