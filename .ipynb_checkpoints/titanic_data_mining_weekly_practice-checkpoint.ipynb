````python
# Import packages
import sqlite3 as sq3
import pandas as pd
import os
````

## 1. Membaca Dataset dengan Pandas

````python
# Lokasi dataset
csv_path = 'data/titanic.csv'

# Membaca CSV
titanic = pd.read_csv(csv_path)

titanic.head()
````

````python
# Melihat ukuran dataset
print('Jumlah baris dan kolom:', titanic.shape)
````

````python
# Melihat nama kolom dan tipe data
titanic.info()
````

## 2. Pemeriksaan Data Awal

Sebelum melakukan analisis, kita perlu mengetahui apakah terdapat nilai yang hilang dan bagaimana karakteristik dasar setiap kolom.


````python
# Jumlah missing value setiap kolom
titanic.isnull().sum()
````

````python
# Statistik deskriptif untuk kolom numerik
titanic.describe()
````

## 3. Membuat Database SQLite

Agar praktik memiliki konsep yang mirip dengan latihan membaca database SQL, data Titanic dimasukkan ke database SQLite terlebih dahulu.


````python
# Membuat folder data jika belum tersedia
os.makedirs('data', exist_ok=True)

# Membuat koneksi SQLite
db_path = 'data/titanic.db'
con = sq3.connect(db_path)

# Menyimpan DataFrame ke tabel SQLite
titanic.to_sql('passengers', con, if_exists='replace', index=False)

print('Database berhasil dibuat:', db_path)
````

````python
# Memastikan tabel tersedia
tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    con
)
tables
````

## 4. Membaca Data dari SQLite

````python
# Membaca seluruh data dari tabel passengers
query = '''
SELECT *
FROM passengers;
'''

observations = pd.read_sql(query, con)
observations.head()
````

## 5. Query SQL untuk Eksplorasi Data

````python
# Jumlah penumpang berdasarkan kelas
query = '''
SELECT Pclass,
       COUNT(*) AS jumlah_penumpang
FROM passengers
GROUP BY Pclass
ORDER BY Pclass;
'''

class_summary = pd.read_sql(query, con)
class_summary
````

````python
# Rata-rata usia berdasarkan jenis kelamin
query = '''
SELECT Sex,
       COUNT(*) AS jumlah_penumpang,
       AVG(Age) AS rata_rata_usia
FROM passengers
GROUP BY Sex;
'''

gender_summary = pd.read_sql(query, con)
gender_summary
````

````python
# Tingkat keselamatan berdasarkan kelas penumpang
query = '''
SELECT Pclass,
       COUNT(*) AS total_penumpang,
       SUM(Survived) AS total_selamat,
       AVG(Survived) AS survival_rate
FROM passengers
GROUP BY Pclass
ORDER BY survival_rate DESC;
'''

survival_summary = pd.read_sql(query, con)
survival_summary
````

## 6. Membandingkan SQL dan Pandas

Analisis yang sama dapat dilakukan menggunakan Pandas `groupby()`.


````python
# Analisis yang sama menggunakan Pandas
pandas_summary = (
    titanic.groupby('Pclass')
    .agg(
        total_penumpang=('PassengerId', 'count'),
        total_selamat=('Survived', 'sum'),
        survival_rate=('Survived', 'mean')
    )
    .reset_index()
)

pandas_summary
````

## 7. Kesimpulan Sementara

Data tabular dapat dibaca menggunakan Pandas maupun database SQLite. SQL memudahkan proses penyaringan dan agregasi data pada tabel, sedangkan Pandas menyediakan cara yang fleksibel untuk eksplorasi dan analisis data di Python.

Pada dataset Titanic, pengelompokan berdasarkan kelas dan jenis kelamin dapat digunakan untuk melihat perbedaan jumlah penumpang, usia rata-rata, dan tingkat keselamatan.
