### Memuat Data California Housing

Dataset dibaca dari `data/california_housing.csv`. Kolom target yang akan diprediksi adalah `MedHouseVal`.

````python
# Muat dataset California Housing dari folder data lokal.
# Struktur folder yang diharapkan:
# project/
# ├── California_Housing_Regularization_LAB.ipynb
# └── data/
#     └── california_housing.csv

import pandas as pd

housing_data = pd.read_csv("data/california_housing.csv")
print("Dataset shape:", housing_data.shape)
print("Columns:", housing_data.columns.tolist())
````

````python
# Tampilkan lima baris pertama dari dataset California Housing
housing_data.head()
````

## Standardisasi Data

**Standardisasi** adalah proses mengubah skala setiap variabel agar memiliki rata-rata 0 dan simpangan baku 1.

Objek [`StandardScaler`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html) dari scikit-learn dapat melakukan proses ini.

**Membuat variabel X dan y:**

````python
y_col = "MedHouseVal"

X = housing_data.drop(y_col, axis=1)
y = housing_data[y_col]
````

**Impor `StandardScaler`, lalu lakukan `fit` dan `transform`:**

````python
from sklearn.preprocessing import StandardScaler

s = StandardScaler()
X_ss = s.fit_transform(X)
````

### Latihan

Pastikan bahwa standardisasi yang dilakukan menghasilkan rata-rata mendekati 0 dan simpangan baku mendekati 1 untuk setiap fitur.

````python
# Petunjuk:

a = np.array([[1, 2, 3], 
              [4, 5, 6]]) 
print(a) # 2 rows, 3 columns
````

````python
a.mean(axis=0) # rata-rata sepanjang kolom
````

````python
a.mean(axis=1) # rata-rata sepanjang baris
````

````python
### AWAL SOLUSI
X2 = np.array(X)
man_transform = (X2-X2.mean(axis=0))/X2.std(axis=0)
np.allclose(man_transform, X_ss)
### AKHIR SOLUSI
````

### Lasso dengan dan tanpa standardisasi

Kita telah membahas Lasso dalam perkuliahan. Mari kita tinjau kembali:

1. Apa perbedaan Lasso dengan Regresi Linear biasa?
2. Apakah standardisasi lebih atau kurang penting untuk Lasso dibandingkan Regresi Linear? Mengapa?

````python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import PolynomialFeatures
````

#### Membuat Fitur Polinomial

Dokumentasi [`PolynomialFeatures`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.PolynomialFeatures.html). Fitur polinomial menambahkan kombinasi pangkat dan interaksi antarfitur.

````python
pf = PolynomialFeatures(degree=2, include_bias=False,)
X_pf = pf.fit_transform(X)
````

**Catatan:** Kita menggunakan `include_bias=False` karena model `Lasso` sudah menyertakan intersep secara bawaan.

````python
X_pf_ss = s.fit_transform(X_pf)
````

### Lasso

Dokumentasi [`Lasso`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Lasso.html).

````python
las = Lasso()
las.fit(X_pf_ss, y)
las.coef_ 
````

### Latihan

Bandingkan hal-hal berikut:

- Jumlah nilai absolut koefisien.
- Jumlah koefisien yang bernilai nol.

Bandingkan model Lasso dengan `alpha=0.1` dan `alpha=1`.

Sebelum menjalankan latihan, jawab masing-masing pertanyaan berikut dalam satu kalimat:

- Model mana yang kamu perkirakan memiliki jumlah nilai absolut koefisien lebih besar?
- Model mana yang kamu perkirakan memiliki lebih banyak koefisien bernilai nol?

````python
### AWAL SOLUSI
las01 = Lasso(alpha = 0.1)
las01.fit(X_pf_ss, y)
print('jumlah nilai absolut koefisien:', abs(las01.coef_).sum() )
print('jumlah koefisien yang tidak nol:', (las01.coef_!=0).sum())
````

````python
las1 = Lasso(alpha = 1)
las1.fit(X_pf_ss, y)
print('jumlah nilai absolut koefisien:',abs(las1.coef_).sum() )
print('jumlah koefisien yang tidak nol:',(las1.coef_!=0).sum())
### AKHIR SOLUSI
````

Dengan regularisasi yang lebih kuat (`alpha` lebih besar), penalti untuk koefisien bernilai besar juga meningkat. Akibatnya, besar koefisien cenderung mengecil dan lebih banyak koefisien dapat terdorong menjadi nol.

### Latihan: Skor $R^2$

Hitung nilai $R^2$ setiap model pada dataset yang sama dengan dataset yang digunakan untuk melatihnya, tanpa pembagian data latih dan data uji.

Fungsi `r2_score` dapat diimpor sebagai berikut:

```python
from sklearn.metrics import r2_score
```

````python
### AWAL SOLUSI
from sklearn.metrics import r2_score
r2_score(y,las.predict(X_pf_ss))
### AKHIR SOLUSI
````

#### Diskusi

## Menggunakan Pembagian Data Latih dan Data Uji

#### Diskusi

Apakah ada masalah dengan pendekatan yang telah kita lakukan sejauh ini?

**Petunjuk:** Pikirkan kembali cara kita melakukan standardisasi fitur. Parameter standardisasi seharusnya dipelajari hanya dari data latih, kemudian diterapkan pada data uji.

Diskusikan dalam kelompok yang terdiri dari dua atau tiga orang.

````python
from sklearn.model_selection import train_test_split
````

````python
X_train, X_test, y_train, y_test = train_test_split(X_pf, y, test_size=0.3, 
                                                    random_state=72018)
````

````python
X_train_s = s.fit_transform(X_train)
las.fit(X_train_s, y_train)
X_test_s = s.transform(X_test)
y_pred = las.predict(X_test_s)
r2_score(y_test, y_pred)
````

````python
X_train_s = s.fit_transform(X_train)
las01.fit(X_train_s, y_train)
X_test_s = s.transform(X_test)
y_pred = las01.predict(X_test_s)
r2_score(y_test, y_pred)
````

### Latihan

#### Bagian 1

Lakukan prosedur yang sama menggunakan Lasso dengan:

- `alpha=0.001`
- `max_iter=100000` untuk membantu memastikan model mencapai konvergensi.

Hitung skor $R^2$ model tersebut.

Kamu boleh menyalin kode dari sel sebelumnya, tetapi tuliskan komentar satu kalimat di atas setiap baris kode untuk menjelaskan tujuannya.

#### Bagian 2

Lakukan prosedur yang sama menggunakan Regresi Linear, lalu hitung skor $R^2$ model tersebut.

#### Bagian 3

Bandingkan jumlah nilai absolut koefisien dan jumlah koefisien bernilai nol dari kedua model. Berdasarkan ukuran tersebut, model mana yang memberikan gambaran hubungan antara fitur dan target yang lebih sederhana?

````python
### AWAL SOLUSI

# Part 1

# Mengurangi regularisasi dan membantu memastikan konvergensi
las001 = Lasso(alpha = 0.001, max_iter=100000)

# Mentransformasi data latih agar terstandardisasi
X_train_s = s.fit_transform(X_train)

# Melatih model menggunakan data latih
las001.fit(X_train_s, y_train)

# Mentransformasi data uji dengan parameter yang dipelajari dari data latih
X_test_s = s.transform(X_test)

# Menghasilkan prediksi pada data uji
y_pred = las001.predict(X_test_s)

# Menghitung skor R²
print("r2 score for alpha = 0.001:", r2_score(y_test, y_pred))


# Part 2

# Using vanilla Linear Regression
lr = LinearRegression()

# Melatih model menggunakan data latih
lr.fit(X_train_s, y_train)

# predicting on test set
y_pred_lr = lr.predict(X_test_s)

# Menghitung skor R²
print("r2 score for Linear Regression:", r2_score(y_test,y_pred_lr))


# Part 3
print('Magnitude of Lasso coefficients:', abs(las001.coef_).sum())
print('Number of coeffients not equal to 0 for Lasso:', (las001.coef_!=0).sum())

print('Magnitude of Linear Regression coefficients:', abs(lr.coef_).sum())
print('Number of coeffients not equal to 0 for Linear Regression:', (lr.coef_!=0).sum())
### AKHIR SOLUSI
````

## Regularisasi L1 dibandingkan L2

Seperti yang dijelaskan dalam materi perkuliahan, penggunaan `Lasso` dan `Ridge` dalam scikit-learn memiliki pola sintaks yang serupa.

Sekarang kita akan membandingkan hasil Ridge dan Lasso.

Dokumentasi [`Ridge`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Ridge.html).

````python
from sklearn.linear_model import Ridge
````

### Latihan

Mengikuti dokumentasi Ridge di atas:

1. Buat objek Ridge bernama `r` dengan nilai `alpha` yang sama seperti `las001`.
2. Latih objek tersebut menggunakan `X` dan `y`, lalu tampilkan koefisien yang dihasilkan.

````python
### AWAL SOLUSI
# Mengurangi regularisasi dan membantu memastikan konvergensi
r = Ridge(alpha = 0.001)
X_train_s = s.fit_transform(X_train)
r.fit(X_train_s, y_train)
X_test_s = s.transform(X_test)
y_pred_r = r.predict(X_test_s)

# Menghitung skor R²
r.coef_
### AKHIR SOLUSI
````

````python
las001 # menggunakan alpha yang sama seperti Ridge di atas
````

````python
las001.coef_
````

````python
print(np.sum(np.abs(r.coef_)))
print(np.sum(np.abs(las001.coef_)))

print(np.sum(r.coef_ != 0))
print(np.sum(las001.coef_ != 0))
````

**Kesimpulan:** Ridge tidak membuat koefisien menjadi nol. Selain itu, pada dataset ini dan untuk nilai `alpha` yang digunakan, Lasso memberikan regularisasi keseluruhan yang lebih kuat daripada Ridge. Hasil ini tidak selalu berlaku untuk semua dataset.

````python
y_pred = r.predict(X_pf_ss)
print(r2_score(y, y_pred))

y_pred = las001.predict(X_pf_ss)
print(r2_score(y, y_pred))
````

**Kesimpulan:** Jika mengabaikan kemungkinan *overfitting*, Ridge sedikit lebih baik daripada Lasso ketika `alpha=0.001` pada dataset ini. Hasil tersebut tidak selalu berlaku secara umum.

# Contoh: Apakah standardisasi memengaruhi hasil?

````python
X_train, X_test, y_train, y_test = train_test_split(X_ss, y, test_size=0.3, 
                                                    random_state=72018)
````

````python
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
r2_score(y_test, y_pred)
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, 
                                                    random_state=72018)
````

````python
s = StandardScaler()
lr_s = LinearRegression()
X_train_s = s.fit_transform(X_train)
lr_s.fit(X_train_s, y_train)
X_test_s = s.transform(X_test)
y_pred_s = lr_s.predict(X_test_s)
r2_score(y_test, y_pred_s)
````

**Kesimpulan:** Untuk Regresi Linear, standardisasi sebelum atau sesudah pembagian data dapat menghasilkan prediksi yang berbeda atau serupa bergantung pada proses yang dilakukan; yang penting, untuk evaluasi yang benar, parameter scaler harus dipelajari dari data latih saja. Pada algoritma lain, scaling dapat sangat berpengaruh. Kita juga dapat memasukkan scaling ke dalam `Pipeline`.
